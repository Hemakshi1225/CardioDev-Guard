# Interactive Heart Disease Predictor

Enter patient metrics below and click **Predict** to see the risk and class prediction.

```python
import joblib
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output

# Load pre-trained logistic regression model
model = joblib.load('logistic_model.pkl')
```

```python
male_widget = widgets.IntText(description='male:', value=0)
age_widget = widgets.FloatText(description='age:', value=0)
education_widget = widgets.IntText(description='education:', value=0)
currentSmoker_widget = widgets.IntText(description='currentSmoker:', value=0)
cigsPerDay_widget = widgets.FloatText(description='cigsPerDay:', value=0)
BPMeds_widget = widgets.IntText(description='BPMeds:', value=0)
prevalentStroke_widget = widgets.IntText(description='prevalentStroke:', value=0)
prevalentHyp_widget = widgets.IntText(description='prevalentHyp:', value=0)
diabetes_widget = widgets.IntText(description='diabetes:', value=0)
totChol_widget = widgets.FloatText(description='totChol:', value=0)
sysBP_widget = widgets.FloatText(description='sysBP:', value=0)
diaBP_widget = widgets.FloatText(description='diaBP:', value=0)
BMI_widget = widgets.FloatText(description='BMI:', value=0)
heartRate_widget = widgets.FloatText(description='heartRate:', value=0)
glucose_widget = widgets.FloatText(description='glucose:', value=0)
```

```python

# Organize layout
inputs = [male_widget, age_widget, education_widget, currentSmoker_widget, cigsPerDay_widget,
          BPMeds_widget, prevalentStroke_widget, prevalentHyp_widget, diabetes_widget,
          totChol_widget, sysBP_widget, diaBP_widget, BMI_widget, heartRate_widget, glucose_widget]
predict_button = widgets.Button(description="Predict")
output = widgets.Output()

def on_predict(b):
    with output:
        clear_output()
        # Gather inputs into DataFrame
        data = {w.description.strip(':'): [w.value] for w in inputs}
        df_input = pd.DataFrame(data)
        # Predict
        prob = model.predict_proba(df_input)[0][1]
        cls = model.predict(df_input)[0]
        print(f"Predicted Class (1 = CHD in 10 yrs): {cls}")
        print(f"Predicted Probability that the patient will develop CHD within ten years: {prob:.2f}")

predict_button.on_click(on_predict)
display(*inputs, predict_button, output)
```
